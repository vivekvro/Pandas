# 14

## pd.crosstab()
### Creates a cross-tabulation (contingency table) of two or more factors, 
### which is useful for summarizing categorical data.

```python
pd.crosstab(df['Time_of_Day'], df['Traffic_Conditions'])
```

# 15 

## .pivot_table() : 
### Similar to pivot(), but allows for aggregation (mean, sum, etc.) of values.

```python
df.pivot_table(index='Traffic_Conditions', columns='Time_of_Day', values='Trip_Price', aggfunc='mean')
```

## .style.backgrund_gradient():
